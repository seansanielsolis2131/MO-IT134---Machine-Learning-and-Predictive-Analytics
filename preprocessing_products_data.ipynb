```python
import pandas as pd
```

```python
file_path = "products_data.csv"
products_data = pd.read_csv(file_path)
```

```python
print(products_data.head())
```

```python
print(products_data.isnull().sum())
```

```python
products_data.loc[:, "Product_Price"] = products_data["Product_Price"].replace({r'[^0-9.]': ''}, regex=True).astype(float)
```

```python
print(products_data.dtypes)
```

```python
print(f"Number of duplicate rows: {products_data.duplicated().sum()}")
```

```python
products_data.to_csv("products_data_cleaned.csv", index=False)
```
